# cavapy: climate data loading

This practical follows one question: How do I easily download future climate projection data for Cuba?

We will load a multi-model dataset and inspect it

## 1. Start the Python kernel

Open this notebook with Python (cavapy). The remote-data cell is the slowest step and requires internet access.

```python
import cavapy

print(cavapy.__version__)
```

## 2. Define the analysis

cavapy can be used to load both observational and projection data

Observational data (ERA5) is available from 1940-2026. The following variables are selected to load observational data:

```python
COUNTRY = "Cuba"
DOMAIN = "CAM-22"
VARIABLE = ["tasmax"]

YEARS_OBS = range(2010,2020)
```

Future projections (CORDEX-CORE) are available from 1980-2005 (historical) and 2006-2100 (future). The following variables are selected to load historical simulations and future projections:

```python
COUNTRY = "Cuba"
DOMAIN = "CAM-22"
DATASET = "CORDEX-CORE-BC"
VARIABLE = ["tasmax"]
RCP = ["rcp26", "rcp85"]
GCM = ["MPI", "MOHC"]
RCM = ["Reg", "REMO"]

YEARS_UP_TO = 2015
```

## 3. Load data

cavapy.get_climate_data() selects Cuba, resolves the remote files, streams the required grid cells and harmonizes units and time.

The following code loads observational data for all variables. Specify the variable (commented out) to only load a single or small number of variables.

```python
obs_data = cavapy.get_climate_data(
    country=COUNTRY,
    obs=True,
  # variable=VARIABLE
    years_obs=YEARS_OBS
)
```

Inspecting the observational data reveals an xarray data structure:

```python
obs_data_tasmax = obs_data["tasmax"]

print(obs_data_tasmax)
```

```python
fig = cavapy.plot_spatial_map(
    obs_data["tasmax"],
    time_period=(2010, 2020),
    title="Mean Max Temperature 2010-2020",
    cmap="Reds",
)

fig = cavapy.plot_time_series(
    obs_data["pr"],
    title="Precipitation Time Series - Cuba (2010-2020)",
    trend_line=True,
    ylabel="Annual Precipitation (mm)",
    aggregation="sum",
    figsize=(12, 6),
)
```

The training uses the pre-bias-corrected CORDEX-CORE archive and so bias-correction is kept as "False". The following code loads historical simulations and future projections:

```python
future_data = cavapy.get_climate_data(
    country=COUNTRY,
    cordex_domain=DOMAIN,
    rcp=RCP,
    gcm=GCM,
    rcm=RCM,
    variables=VARIABLE,
    years_up_to=YEARS_UP_TO,
    historical=True,
    bias_correction=False,
    dataset=DATASET,
)
```
